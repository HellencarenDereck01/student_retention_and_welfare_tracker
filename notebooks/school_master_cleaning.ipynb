```python
# STEP 5: DATA CLEANING
```

```python
clean_df = df.copy()

print("Cleaning DataFrame created.")
print("Rows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])
```

```python
text_columns = clean_df.select_dtypes(include="object").columns

for column in text_columns:
    clean_df[column] = clean_df[column].replace(r"^\s*$", np.nan, regex=True)

print("Blank text values have been converted to missing values.")
```

```python
missing_after_blank_check = clean_df.isnull().sum()

print("Missing values after checking blank entries:")
print(missing_after_blank_check)
```

```python
clean_df["district"] = clean_df["district"].fillna("Unknown")
clean_df["block"] = clean_df["block"].fillna("Unknown")

print("Missing values in district and block have been corrected.")
```

```python
print("Missing values after correction:")
print(clean_df.isnull().sum())
```

```python
text_columns = clean_df.select_dtypes(include="object").columns

for column in text_columns:
    clean_df[column] = (
        clean_df[column]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

print("Extra spaces have been removed from text columns.")
```

```python
for column in text_columns:
    print(f"\nColumn: {column}")
    print(clean_df[column].head(5).tolist())
```

```python
duplicate_count = clean_df.duplicated().sum()

print("Duplicate rows before removal:", duplicate_count)

clean_df = clean_df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows after removal:", clean_df.duplicated().sum())
print("Rows remaining:", len(clean_df))
```

```python
duplicate_school_ids = clean_df["school_id"].duplicated().sum()

print("Number of duplicate school IDs:", duplicate_school_ids)
```

```python
print(clean_df.dtypes)
```

```python
clean_df["total_enrolled_students"] = pd.to_numeric(
    clean_df["total_enrolled_students"],
    errors="coerce"
)

print(clean_df["total_enrolled_students"].dtype)
```

```python
negative_student_totals = (
    clean_df["total_enrolled_students"] < 0
).sum()

print("Number of negative student totals:", negative_student_totals)
```

```python
print(
    "Missing values in total_enrolled_students:",
    clean_df["total_enrolled_students"].isnull().sum()
)
```

```python
columns_to_format = ["district", "block"]

for column in columns_to_format:
    clean_df[column] = (
        clean_df[column]
        .astype("string")
        .str.strip()
        .str.lower()
        .str.title()
    )

print("district and block have been formatted.")
```
