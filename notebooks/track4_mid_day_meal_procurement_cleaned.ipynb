```python
# STEP 5:DATA CLEANING
```

```python
# Create a copy so the original raw data remains unchanged
clean_df = df.copy()

for column in clean_df.select_dtypes(include="object").columns:
    clean_df[column] = clean_df[column].str.strip()

missing_values = ["", "NA", "N/A", "na", "n/a", "null", "None", "-"]

clean_df = clean_df.replace(missing_values, np.nan)

clean_df.head()
```

```python
# Create a cleaning copy
clean_df = df.copy()
```

```python
# Remove duplicate rows
clean_df = clean_df.drop_duplicates()

# Remove extra spaces from text columns
for col in clean_df.select_dtypes(include="object").columns:
    clean_df[col] = clean_df[col].str.strip()

print("Duplicates remaining:", clean_df.duplicated().sum())
```

```python
# Standardize school_id
import re

def clean_school_id(value):
    if pd.isna(value):
        return value

    number = re.search(r"\d+", str(value))

    if number:
        return "SCH" + number.group().zfill(4)

    return np.nan

clean_df["school_id"] = clean_df["school_id"].apply(clean_school_id)

clean_df["school_id"].head()
```

```python
# Standardize the date format
clean_df["date"] = pd.to_datetime(
    clean_df["date"],
    format="mixed",
    errors="coerce",
    dayfirst=True
)

clean_df["date"] = clean_df["date"].dt.strftime("%Y-%m-%d")

clean_df["date"].head()
```

```python
print("Invalid dates:", clean_df["date"].isna().sum())
```

```python
# Standardize vendor names
clean_df["vendor_name"] = (clean_df["vendor_name"].str.lower().str.strip())
vendor_map = {
    "sharma traders pvt ltd": "Sharma Traders",
    "sharma traders private limited": "Sharma Traders",
    "sharma trader": "Sharma Traders",
    "sharma traders": "Sharma Traders",

    "singh brother": "Singh Brothers",
    "singh brothers": "Singh Brothers",

    "kumar and co": "Kumar & Co.",
    "kumar co": "Kumar & Co.",

    "kumar general store": "Kumar General Store",
    "kumar supplies": "Kumar Supplies",
    "goyal rice mill": "Goyal Rice Mill"
}

clean_df["vendor_name"] = clean_df["vendor_name"].replace(vendor_map)

clean_df["vendor_name"].head()
```

```python
# Standardize grain names
grain_map = {
    "rice": "Rice",
    "chawal": "Rice",

    "wheat": "Wheat",
    "gehun": "Wheat",

    "dal": "Lentils",
    "daal": "Lentils",
    "lentils": "Lentils",

    "pulses": "Pulses",
    "pulse": "Pulses",

    "sarson": "Mustard",
    "mustard": "Mustard"
}

clean_df["grain_type"] = (
    clean_df["grain_type"]
    .str.lower()
    .replace(grain_map)
)

clean_df["grain_type"].head()
```

```python
# Convert all quantities to kilograms
def convert_to_kg(value):
    if pd.isna(value):
        return np.nan

    value = str(value).lower().replace(",", "")

    number = re.search(r"\d+(?:\.\d+)?", value)

    if not number:
        return np.nan

    number = float(number.group())

    if "gram" in value or re.search(r"\bg\b", value):
        return number / 1000

    if "bag" in value or "sack" in value or "bori" in value:
        return number * 50

    return number
```

```python
# Clean total_cost
clean_df["total_cost"] = (
    clean_df["total_cost"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+(?:\.\d+)?)")[0]
)

clean_df["total_cost"] = pd.to_numeric(
    clean_df["total_cost"],
    errors="coerce"
)

clean_df["total_cost"].head()
```

```python
for column in ["total_cost"]:
    clean_df[column] = clean_df[column].fillna(clean_df[column].median())
```

```python
# Standardize payment status
payment_map = {
    "paid": "Paid",
    "cleared": "Cleared",
    "pending": "Pending",
    "due": "Due",
    "unpaid": "Unpaid"
}

clean_df["payment_status"] = (
    clean_df["payment_status"]
    .str.lower()
    .replace(payment_map)
)

clean_df["payment_status"].head()
```

```python
categorical_columns = [
    "vendor_name",
    "grain_type",
    "unit",
    "payment_status"
]

for col in categorical_columns:
    clean_df[col] = clean_df[col].fillna(
        clean_df[col].mode()[0]
    )
```

```python
import re
import numpy as np
import pandas as pd

def convert_to_kg(value):
    if pd.isna(value):
        return np.nan

    value = str(value).lower().strip()
    value = value.replace(",", "")

    # Get the number from the value
    number_match = re.search(r"\d+(?:\.\d+)?", value)

    if number_match is None:
        return np.nan

    number = float(number_match.group())

    # Convert grams to kilograms
    if "gram" in value or re.search(r"\bg\b", value):
        return number / 1000

    # Convert bags, sacks, and boris to kilograms
    elif "bag" in value or "sack" in value or "bori" in value:
        return number * 50

    # KG or no unit: treat as kilograms
    else:
        return number
```

```python
clean_df["quantity_kg"] = clean_df["quantity"].apply(convert_to_kg)

display(clean_df[["quantity", "unit", "quantity_kg"]].head(20))

print("Missing quantity values:", clean_df["quantity_kg"].isna().sum())
clean_df["quantity_kg"] = clean_df["quantity_kg"].fillna(
    clean_df["quantity_kg"].median()
)
clean_df["quantity"] = clean_df["quantity_kg"]

clean_df = clean_df.drop(columns=["quantity_kg"])
```

```python
# Make a copy of the cleaned data
clean_df = clean_df.copy()

# Standardize all unit values to KG
clean_df["unit"] = "KG"

# Check the result
print(clean_df["unit"].unique())

display(clean_df[["quantity", "unit"]].head(20))
```
