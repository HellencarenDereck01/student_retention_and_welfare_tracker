## Step 1 — Test Score Standardization

The test-score dataset contains multiple grading scales, including:

- Percentage (`%`, `pct`, `Percentage`)
- Raw Marks
- CGPA
- Letter Grade

To make scores comparable across schools, grades, and subjects, all scores are
standardized to a percentage scale from 0 to 100.

The original `avg_score` and `grading_scale` columns are retained for traceability,
while a new `score_percentage` column is created for analysis.

```python
raw_marks = test_scores.loc[
    test_scores["grading_scale"] == "Raw Marks",
    "avg_score"
]

raw_marks.head(20).tolist()
```

```python
test_scores_clean = test_scores.copy()
```

```python
def convert_to_percentage(row):
    scale = row["grading_scale"]
    score = row["avg_score"]
    
    # Percentage-based scores
    if scale in ["%", "pct", "Percentage"]:
        return float(str(score).replace("%", "").strip())
    
    # CGPA scores
    elif scale == "CGPA":
        return float(score) * 9.5
    
    # Raw marks such as 21.6/50 or 16.0/25
    elif scale == "Raw Marks":
        score_value, max_value = str(score).split("/")
        return (float(score_value) / float(max_value)) * 100
    
    # Letter grades
    elif scale == "Letter Grade":
        grade_mapping = {
            "A+": 95,
            "A": 85,
            "B": 75,
            "C": 65,
            "D": 55,
            "E": 45
        }
        return grade_mapping.get(str(score).strip())
    
    return np.nan
```

```python
test_scores_clean["score_percentage"] = test_scores_clean.apply(
    convert_to_percentage,
    axis=1
)
```

## Step 2 — Date Standardization

The raw test-score dataset contains dates in multiple formats, including:

- `YYYY-MM-DD`
- `YYYY/MM/DD`
- `DD.MM.YYYY`
- `DD/MM/YYYY`
- `DD-MM-YYYY`
- `MM-DD-YYYY`
- `DD-MMM-YYYY`

These formats are converted into a single standardized date format.

The cleaned dates are stored in the `date_clean` column using the
`YYYY-MM-DD` format.

Where numeric dates are ambiguous, the structure of the date is examined
carefully to avoid incorrectly swapping the day and month values.

```python
test_scores_clean["date_clean"] = pd.to_datetime(
    test_scores_clean["date"],
    errors="coerce",
    dayfirst=False
)
```

```python
def parse_date(value):
    value = str(value).strip()

    # YYYY-MM-DD or YYYY/MM/DD
    if len(value) >= 10 and value[4] in ["-", "/"]:
        return pd.to_datetime(value, errors="coerce", yearfirst=True)

    # DD.MM.YYYY
    if "." in value:
        return pd.to_datetime(
            value,
            format="%d.%m.%Y",
            errors="coerce"
        )

    # DD/MM/YYYY
    if "/" in value:
        return pd.to_datetime(
            value,
            format="%d/%m/%Y",
            errors="coerce"
        )

    # Dates containing month names, e.g. 06-Mar-2026
    if any(char.isalpha() for char in value):
        return pd.to_datetime(
            value,
            format="%d-%b-%Y",
            errors="coerce"
        )

    # Numeric dates containing hyphens
    if "-" in value:
        parts = value.split("-")

        if len(parts) == 3:
            first = int(parts[0])
            second = int(parts[1])

            # MM-DD-YYYY, e.g. 10-20-2025
            if second > 12:
                return pd.to_datetime(
                    value,
                    format="%m-%d-%Y",
                    errors="coerce"
                )

            # DD-MM-YYYY, e.g. 06-11-2025
            return pd.to_datetime(
                value,
                format="%d-%m-%Y",
                errors="coerce"
            )

    return pd.NaT


test_scores_clean["date_clean"] = test_scores_clean["date"].apply(parse_date)
```

## Step 3 — School ID Standardization

The `school_id` column contains multiple representations of the same school,
such as:

- `SCH1001`
- `SCH-1001`
- `sch_1001`
- `S1001`
- `1001`

These variations can prevent accurate joins between datasets.

To ensure consistent joining across the project, all school identifiers are
standardized to the canonical format:

`SCH####`

The original `school_id` column is retained and a new `school_id_clean`
column is created for the standardized identifier.

```python
test_scores_clean["school_id"].value_counts().head(20)
```

```python
print("Unique school IDs:", test_scores_clean["school_id"].nunique())
```

```python
print(
    test_scores_clean["school_id"]
    .astype(str)
    .head(20)
    .tolist()
)
```

```python
def standardize_school_id(value):
    value = str(value).strip().upper()

    # Remove common separators
    value = value.replace("-", "").replace("_", "")

    # Remove SCH prefix if present
    if value.startswith("SCH"):
        value = value[3:]

    # Remove S prefix if present
    elif value.startswith("S"):
        value = value[1:]

    # Keep the numeric part and restore leading zeros
    if value.isdigit():
        return "SCH" + value.zfill(4)

    return np.nan


test_scores_clean["school_id_clean"] = (
    test_scores_clean["school_id"].apply(standardize_school_id)
)
```

## Step 4 — Subject Standardization

The raw dataset contains multiple names that refer to the same subject.

In particular:

- `Ganit` → `Math`
- `Mathematics` → `Math`
- `Math` → `Math`

The remaining subjects are retained as they appear in the dataset.

This reduces duplicate subject categories and ensures that mathematics
assessments are analyzed together.

A new `subject_clean` column is created while the original `subject` column
is retained for traceability.

```python
test_scores_clean["subject"].value_counts()
```

```python
subject_mapping = {
    "Ganit": "Math",
    "Mathematics": "Math"
}

test_scores_clean["subject_clean"] = (
    test_scores_clean["subject"]
    .replace(subject_mapping)
)
```

## Handling Placeholder Values

The raw dataset uses string placeholders such as "NA" instead of actual missing values.
These placeholders are converted into proper missing values (`NaN`) for consistent analysis.

```python
# Count placeholder values before cleaning

print("NA values in max_marks:")
print((test_scores_clean["max_marks"] == "NA").sum())
```

```python
test_scores_clean["max_marks"] = (
    test_scores_clean["max_marks"]
    .replace("NA", np.nan)
)
```

```python
print("Missing values in max_marks:")
print(test_scores_clean["max_marks"].isna().sum())
```

## Missing Value Summary

After cleaning placeholder values, we inspect the dataset for remaining missing values.

```python
missing_summary = (
    test_scores_clean.isna()
    .sum()
    .sort_values(ascending=False)
)

missing_summary
```

### Cleaning Decision

`max_marks` is missing only for records using the **Letter Grade** grading scale.

Since Letter Grade assessments do not use numerical maximum marks, these missing values are **structural missing values**, not data quality errors.

The values are intentionally kept as `NaN`.
